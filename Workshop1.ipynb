```python
Lena_gray = color_lena.convert("L")
display(Lena_gray)
Lena_gray_arr = np.array(Lena_gray)
print(Lena_gray_arr.shape)
height, weight = Lena_gray_arr.shape
data = Lena_gray_arr.reshape(height,weight)
```

```python
mean = np.mean(data,axis =0)
centered_data = data - mean
print(centered_data)
```

```python
cov_matrix = np.cov(centered_data)
print(cov_matrix)
```

```python

eigenvalues, eigenvectors = np.linalg.eigh(cov_matrix)
```

```python
#sort in descending order
sort = np.argsort(eigenvalues)[::-1]
eigenvalues = eigenvalues[sort]
eigenvectors = eigenvectors[:,sort]
```

```python
explained_variance_ratio = eigenvalues / np .sum ( eigenvalues )
plt . plot ( np . cumsum ( explained_variance_ratio ) )
plt . title (" Cumulative Explained Variance ")
plt . xlabel (" Number of Components ")
plt . ylabel (" Cumulative Explained Variance ")
plt . grid ( True )
plt . show ()
```

```python
a = 10
b=20
c=50
d=100
e=150
components1 = eigenvectors[:, :a]
components2 = eigenvectors[:, :b]
components3 = eigenvectors[:, :c]
components4 = eigenvectors[:, :d]
components5 = eigenvectors[:, :e]
```

```python
compressed_data1 = np.dot(centered_data, components1)
compressed_data2 = np.dot(centered_data, components2)
compressed_data3 = np.dot(centered_data, components3)
compressed_data4 = np.dot(centered_data, components4)
compressed_data5 = np.dot(centered_data, components5)

print(centered_data.shape)
print(components1.shape)
print(components2.shape)
print(components3.shape)
print(components4.shape)
print(components5.shape)
print(compressed_data1.shape)
print(compressed_data2.shape)
print(compressed_data3.shape)
print(compressed_data4.shape)
print(compressed_data5.shape)
```

```python
decompressed_data1 = np.dot(compressed_data1, components1.T) + mean
decompressed_data2 = np.dot(compressed_data2, components2.T) + mean
decompressed_data3 = np.dot(compressed_data3, components3.T) + mean
decompressed_data4 = np.dot(compressed_data4, components4.T) + mean
decompressed_data5 = np.dot(compressed_data5, components5.T) + mean
print(decompressed_data1.shape)
print(decompressed_data2.shape)
print(decompressed_data3.shape)
print(decompressed_data4.shape)
print(decompressed_data5.shape)
```
